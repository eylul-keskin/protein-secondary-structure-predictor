# Keşifsel İnceleme ve Güncel Sonuçlar

Bu hafif notebook, `outputs/` altındaki veri kontrolü ve mevcut checkpoint'in yeniden değerlendirme çıktılarını okur. Eğitim veya model tahmini başlatmaz.

Güncel tablolar ve karışıklık matrisleri yeni bir eğitim koşusu olarak sunulmaz. Bu checkpoint ile uyumlu eğitim eğrisi bulunmadığından eğitim geçmişi görseli beklenmez. Uyumsuz eski üç epokluk kayıtlar, açıklamalarıyla `archive/legacy_results/` altında ayrılmıştır.



```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, display

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
OUTPUTS = ROOT / 'outputs'
OUTPUTS
```

```python
validation_summary = json.loads((OUTPUTS / 'logs' / 'data_validation_summary.json').read_text(encoding='utf-8'))
schema_summary = pd.DataFrame([validation_summary['schema']])
row_summary = pd.DataFrame(
    [
        {
            'split': split_name,
            'rows': split_data['total_rows'],
            'removed_rows': split_data['removed_rows'],
            'duplicate_rows': split_data['duplicate_rows'],
            'min_length': split_data['sequence_length_stats']['min'],
            'max_length': split_data['sequence_length_stats']['max'],
            'mean_length': split_data['sequence_length_stats']['mean'],
        }
        for split_name, split_data in validation_summary['split_validation'].items()
    ]
)
display(schema_summary)
display(row_summary)
```

```python
comparison = pd.read_csv(OUTPUTS / 'metrics' / 'model_comparison.csv')
test_comparison = pd.read_csv(OUTPUTS / 'metrics' / 'test_set_comparison.csv')
display(comparison)
display(test_comparison)
```
